# Exception Handling

* An exception is an error that occurs while a Python program is running.
* when an error occurs, Python stops the program and displays an error message.
* Exception handling allows us to detect the error and handle it properly instead of suddenly stopping the program.

* try → contains code that may cause an error.
* except → handles the error.
* else → runs when no error occurs.
* finally → runs whether an error occurs or not.
* raise → manually creates an exception.
* assert → checks whether a condition is true.
* Custom Exception → allows us to create our own exception types.

### Real time example

* we are trying to withdraw money.

### example for try

```python
try:
    number = int(input("Enter a number: "))
    print("Number:", number)

except ValueError:   
    print("Please enter a valid number.")
```

### except
* hanldes the error occured in try block
### real time example
* if there is insufficient balance,the Atm handles the situation

```python
try:
    number = int(input("Enter a number: "))
    print("Number:", number)

except ValueError:   
    print("Please enter a valid number.")
```

### else
* The else block executes only when the try block does not generate an exception.
### real time example
* If the withdrawal is successful, the ATM gives you the money.

```python
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Invalid input.")
else:
    print("You entered:", number)
```

### finally
* The finally block executes whether an exception occurs or not.

### real time example
* The ATM ends the transaction and returns to the main screen.

```python
try:
    number = int(input("Enter a number: "))
    print(number/5)
except valueError:
    print("Enter a valid number")
finally:
    print("Program  completed.")
```

## raise
* it is used when we want to manually generate an exception.

### real time example
* The ATM can reject an invalid transaction.

### example

```python
age = int(input("Enter your age: "))
if age < 0:
    raise ValueError("Age cannot be negative.")
print("Age:", age)
```

# assert
* checks whether a condition is true.

### real time example
* The system checks that certain conditions are valid.

### example


```python
marks = 50
assert marks >= 80
print("Valid marks")
```

# Custom Exception
* created by the programmer for a specific situation.

### real time example
*  Custom Exception → A bank can create its own error such as InsufficientBalanceError.

### example

```python
class InsufficientBalanceError(Exception):
    pass
bal = 106
if bal < 1000:
    raise InsufficientBalanceError("Insufficient balance")
```

# Types of errors

### Value Error

* occurs when a function receives the correct type of value but an inappropriate value.

### example

```python
try:
    num = int(input("Enter a number: "))
    print("Number:", num)
except ValueError:
    print("Invalid number")
```

### Type Error
* occurs when an operation is performed on an incompatible data type.
### example

```python
try:
    number = 10
    text = "20"
    result = number + text
    print(result)
except TypeError:
    print("Cannot add an integer and a string.")
```

### Index error
* occurs when we try to access a list or sequence position that does not exist.
### example

```python
try:
    numbers = [10, 20, 30]
    print(numbers[5])

except IndexError:
    print("The specified index does not exist.")
```

### Key error
* occurs when we try to access a dictionary using a key that does not exist.

### example

```python
try:
    student = {
        "name": "Ravi",
        "marks": 85
    }
    print(student["age"])
except KeyError:
    print("The requested key does not exist.")
```

### Attribute error
* occurs when an object does not have the attribute or method we try to access.
### example

```python
try:
    name = "Python"
    name.append("Programming")   # gets error here because of string does not contain append method
except AttributeError:
    print("The object does not have this method.")
```

### Name error
* occurs when Python cannot find a variable or name that has not been defined.
### example

```python
try:
    print(student_name) # student_name is not defined in this program so here we will get NameError
except NameError:
    print("The variable has not been defined.")
```

### FileNotFoundError
* occurs when we try to open a file that does not exist.

### example

```python
try:
    file = open("student.txt", "r")  #  there is no file named student.txt so we gegt FileNotFoundError
    print(file.read())
    file.close()
except FileNotFoundError:
    print("The requested file was not found.")
```

### ZeroDivisionError
* when we try to divide a number by zero then it occurs ZeroDivisionError
### example

```python
try:
    result = 10 / 0  # here 10 is dividing with zero so it occurs ZeroDivisionError
    print(result)

except ZeroDivisionError:
    print("Cannot divide a number by zero.")
```

### ImportError
* occurs when Python cannot import the requested name or component from a module.

### example

```python
try:
    from math import student
except ImportError:
    print("The requested item could not be imported.")
```

### ModuleNotFoundError
* occurs when Python cannot find the module that we are trying to import.

### example

```python
try:
    import my_unknown_module 
except ModuleNotFoundError:
    print("The requested module was not found.")
```
