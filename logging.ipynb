# Print() vs logging

* both print and logging displays information
* prints the specified value or values to the console.
* but  logging is used to record a message to the console.

```python
import logging
logger = logging.getLogger()
logger.warning('This is a Warning message!')   # setting logger to a warning message
logging.basicConfig(level=logging.INFO)     # configuring the logger to info log levek
logging.info("This is an Informative message.")  # setting the logger to a informative message
```

# log levels
* levels always works in heirarchy
* default logging level is always warning

## DEBUG
* It is used to record simple values
## INFO
*  it is used to record general information
## WARNING
* it shows the issues which indicates something is not right but does not stop execution
## ERROR
* it records errors which causes a section of code to fail
## CRITICAL
* It is the highest level,which fails the whole program
* 0 represents that level is not set

### Implementation of logging
* at first we have to create a log file to store messages 
* we have to specify which level of messages are to be stored using basicConfig() function using logging module
* after creating a logfile we can write messages to that file by usinf following methods

```python
# logging.critical(msg)
# logging.error(msg)
# logging.warning(msg)
# logging.info(msg)
# logging.debug(msg)
```

```python
import logging
logging.basicConfig(filename='MyApp.log')
logging.debug('this is debug message')
logging.info('this is information message')
logging.warning('this is warning message')
logging.error('this is error message')
logging.critical('this is a critical message')
```

## Log Formatters

* It controls how a log message appears
### real time example
2026-09-25 09:30:10 - INFO - Application started

```python
import logging

logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s - %(levelname)s - %(message)s"   # %(asctime)s-->defines time,%(levelname)s-->defines log level,%(message)s-->defines log message
)

logging.info("Application started")
logging.warning("This is a warning")
```

## File Handlers
* A FileHandler sends log messages to a file instead of only displaying them on the console.
* it sends logs to a file so they can be saved and checked later.

```python
import logging
logger = logging.getLogger("MyApp")                  # Create logger
logger.setLevel(logging.INFO)

file_handler = logging.FileHandler("app.log")          # Create FileHandler

logger.addHandler(file_handler)                       # Connect handler to logger
logger.info("Application started")
logger.warning("This is a warning")
logger.error("An error occurred")
```

## Console handler

* used in Python logging to display log messages on the console/terminal while the program is running.
### real time example
* Imagine a student management application.

When the application is running, the developer wants to see important events immediately:

Application started /
Student record added /
Warning about low marks /
Error while processing a record
* These messages can be displayed directly in the console using a Console Handler.

```python
import logging
logger = logging.getLogger("StudentApp")
logger.setLevel(logging.INFO)
console_handler = logging.StreamHandler()   # Create Console Handler
logger.addHandler(console_handler)           # Add handler to logger

logger.info("Student application started")
logger.warning("Student marks are low")
logger.error("Unable to process student record")
```

## Rotating logs

*  In a large application, log files can become very large.
* RotatingFileHandler allows us to automatically create a new log file after the current file reaches a specified size.

### real time example
* Imagine a bike showroom. If the showroom already has many bikes and the storage area becomes full, 
* the owner can create a new storage shed and move older bikes there. 
* This allows the showroom to continue storing new bikes without letting one storage area become too large.

### example
* Bike showroom --> Application
* Bikes --> Log messages
* Storage shed --> Log file
* Storage becomes full --> Log file reaches its maximum size
* New storage shed --> New log file
* Older bikes moved to previous shed --> Older logs are kept in backup files

```python
import logging
from logging.handlers import RotatingFileHandler
logger = logging.getLogger("BankApp")
logger.setLevel(logging.INFO)
handler = RotatingFileHandler(              # Create rotating file handler
    "bank.log",
    maxBytes=100,
    backupCount=2
)
logger.addHandler(handler)               # Add handler to logger
logger.info("Bank application started")
logger.warning("Low account balance")
logger.error("Transaction failed")
```

# Logging Best Practices

#### 1.use appropriate log levels
 *   1.DEBUG
  *  2.INFO
   * 3.ERROR
   * 4.WARNING
   * 5.CRITICAL

#### 2. use meaningful messages
* without writing simply an error instead write what type of error

#### 3. Don't log sensitive information
* like passwords or credit card numbers or authentication code....

#### 4.use logging instead of print()
* For a small beginner program, print() is fine.
* For larger applications, logging provides better control and debugging capabilities.

#### 5. Configure logging centrally
* A larger application should have a consistent logging configuration instead of configuring logging differently in every file.
