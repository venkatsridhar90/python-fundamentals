# Module-2: Object-Oriented Programming

## 1.Class

### what is class?


* A class is a blueprint for creating objects. It doesn't hold real data by itself
* it defines what shape the data will take (attributes) and what behaviors are possible (methods).
*  Think of it as a template nothing exists until you actually build something from it

### Real time example

* A class is like an architectural blueprint for a house. The blueprint specifies 2 bedrooms, 1 kitchen but you can't live in a blueprint.
* so we need to  construct a house from the blueprint. Many houses (objects) can be built from one blueprint (class).

### example

```python
class Student:    #  creates a class
    name = "John"  # attributs of a class
    age = 20      

student1 = Student() 

print(student1.name)
print(student1.age)
```

# 2.Object

An Object is an instance of a Class. It represents a specific implementation of the class and holds its own data. 
* State: represented by the attributes and reflects the properties of an object.
* Behavior: represented by the methods of an object and reflects the response of an object to other objects.

## Real time example
* Imagine a printing press producing identity cards using one card template.
* The template represents the Class, while each printed ID card is an Object.
* Every card has the same fields like a photo slot, a name field, an ID number , but each card ends up holding a different person's photo, name, and number once printed.
* Similarly, every object follows the structure defined by the class while filling that structure with its own values.

### Example

```python
class Student:
    name = "Ravi"
    age = 20
student1 = Student() # Creating an object
print(student1.name) #prints name
print(student1.age)
```

# Constructor (__init__)

A constructor is a special method that runs automatically when an object is created.
* Its job is to set up the object's initial state ,assigning starting values to its attributes.

### Real time example
* Imagine a restaurant kitchen that automatically preheats the oven, sharpens the knives, and lays out the ingredients
*  the moment a chef starts a new shift — before a single dish is cooked. The chef never begins working from a completely blank, unprepared station.
*  Similarly, a constructor automatically prepares an object's initial data the moment it's created, before any of its methods are called.
*  This analogy demonstrates that a constructor guarantees every object begins life in a complete, ready-to-use state.

# example:


```python
class Student:
    def __init__(self, name, age): # automatically runs when student("Sridhar",20) is executed
        self.name = name
        self.age = age
    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
student1 = Student("Sridhar", 20)
student1.display()
```

# 4.Instance Variables:
* Instance variables are variables that belong to a particular object.
* Different objects can have different values.

### Real time example
* Every student has their own backpack.
* Filling your backpack with books doesn't fill your classmate's backpack , each backpack belongs only to that one student (object).

### Example

```python
class Student:
    def __init__(self, name, grade):
        self.name = name      # instance variable
        self.grade = grade    # instance variable

s1 = Student("Ravi", "A")
s2 = Student("Sridhar", "B")
print(s1.name, s1.grade)
print(s2.name,s2.grade)
```

# 5.Class Variables
* A class variable is shared by all objects of a class.
* it lives on the class itself, not on any single instance

### real time example
* A school's current academic year(e.g., 2026) applies to every student equally,it's not something each student sets individually.
* Update it once at the school level, and it applies to all students, just like a class variable.

### example

```python
class Student:
    school_name = "Balaji High School"  # class variable

    def __init__(self, name):
        self.name = name  # instance variable

s1 = Student("Ravi")
s2 = Student("Sridhar")
print(s1.name, s2.school_name)
print(s2.name,s2.school_name)
```

# 6.Instance methods

* An instance method is a method that works with the data of a particular object.
* They always take self as their first parameter

### real time example
* A student can display their own marks.
* The introduction depends on the particular student's data.

### example


```python
class Student:
    def __init__(self, name, marks):
        self.name = name      # Store the student's name
        self.marks = marks    # Store the student's marks
    def display(self):
        print(self.name, self.marks)
student1 = Student("Ravi", 85) # Create an object of the Student class
# Call the display method using the object
student1.display()
```

# 7.static methods
* A static method belongs to the class but does not need object-specific data.
* It behaves like a normal function but is logically grouped inside a class.

### real time example
* A calculator's addition operation does not need student information.

### example

```python
class Calculator:
    @staticmethod             # Define a static method
    def add(a, b):              # Method accepts two numbers
        return a + b      
result = Calculator.add(45, 20)     # Call the static method using the class name
print(result)
```

# 8.class methods
* A class method works with the class itself rather than a particular object.
* It's commonly used for operations that affect the class as a whole or as alternative constructors that build objects in a different way.

### real time example
* A passport office's issue a new passport from an application form process operates on the Passport concept in general,
* not on one specific passport that already exists
*  it's a class-level operation that reads the application and produces a brand-new passport instance

### example

```python
class Student:
    college = "VIT UNIVERSITY"
    @classmethod   # Define a class method
    def display_college(cls):   # Method receives the class as cls
        print(cls.college)
Student.display_college() # Call the class method using the class name
```

# 9.properties(@property)
* A property allows us to access a method like a normal variable.
* It is useful when we want to control how a value is accessed or modified.

### real time example
* A thermostat displays "72°F" as a simple reading, but internally it's constantly running sensor logic to compute that number.
* Similarly, a property can validate a new value before storing it, silently rejecting anything that would leave the object in an invalid state.


### example

```python
class Student:
    def __init__(self, marks):
        self.marks = marks
    @property # Create a property named percentage
    def percentage(self):
        return self.marks
student = Student(85)
print(student.percentage) ## Access the percentage property
```

# 10.Encapsulation
* Encapsulation means bundling data and the methods that operate on it together, while restricting direct outside access to some of that data
* It also allows us to control direct access to internal data.

### real time example
* A bank vault doesn't let customers walk in and grab cash directly.
* Instead, they interact through a teller (a controlled method) who enforces rules like withdrawal limits.
* The cash (data) is encapsulated behind that controlled interface.

### example

```python
class BankAccount:
    def __init__(self, balance):
        self.bal = balance   # Store balance as a private variable
    def deposit(self, amount):
        self.bal += amount # Add the deposited amount to the balance
    def show_balance(self):
        print("Balance:", self.bal)
account = BankAccount(5000)   # Create an object with an initial balance of 5000
account.deposit(1000)  # Deposit 1000 into the account
account.show_balance()
```

# 11.Abstraction
* Abstraction means showing only the necessary details while hiding the internal implementation.
* an abstract base class declares methods that subclasses must implement, without specifying the internal details.

### real time example
* Using a washing machine, you just select a cycle and press start you don't need to know how the motor drives the drum,
*  how water levels are sensed, or how the timing circuits sequence each phase.
*  The machine abstracts away all that mechanical and electronic complexity behind a few buttons and a dial.

### example

```python
from abc import ABC, abstractmethod
class Vehicle(ABC):
    @abstractmethod    # Define an abstract method
    def start(self):
        pass            # Method body is defined in the child class
class Car(Vehicle):
    def start(self):
        print("Car starts with a key")
car = Car()              # Create an object of the Car class
car.start()
```

# 12.Inheritance
* Inheritance allows one class to reuse properties and methods from another class.
* The existing class is called the parent/base class.
* The new class is called the child/derived class.

### real time example

* A mobile phone can be a parent class, while Android phones or specific types of mobile phones can be child classes that inherit its common features.
* mobile phone --> parent class
* android mobile  -->child class

### example

```python
class Animal:
    def eat(self):
        print("Animal eats")
class Dog(Animal):  # Dog class is acquiiring all properties of an animal
    def bark(self):
        print("Dog barks")
dog = Dog()
dog.eat()
dog.bark()
```

# 13.Multiple Inheritance
* Multiple inheritance means a class inherits from more than one parent class.

### real time example
* A smartphone can have features of:
*     Camera
*     Phone

### example

```python
class Camera:

    def take_photo(self):  # Method to take a photo
        print("Taking photo")
        
class Phone:              # Create another parent class named Phone
    def make_call(self):    # Method to make a call
        print("Making call")
        
class Smartphone(Camera, Phone):     # Create a child class that inherits from both Camera and Phone
    pass
mobile = Smartphone()

mobile.take_photo()
mobile.make_call()
```

# 14.Multi level Inheritance
* Multilevel inheritance occurs when inheritance happens in multiple levels.

### real time example
* Grandparent → Parent → Child.
* The child inherits traits from the parent, who in turn inherited traits from the grandparent

### example

```python
class Person:  # Parent class
    def introduce(self):  
        print("I am a person")  

class Employee(Person):  
    def work(self):  
        print("I work in a company") 
        
class Manager(Employee):  # Manager inherits from Employee
    def manage(self):  
        print("I manage a team")  
manager = Manager()  # Create an object of Manager class

manager.introduce()  
manager.work()      
manager.manage()
```

# 15.Hierarchical Inheritance
* Hierarchical inheritance means multiple child classes inherit from the same parent.
* cat<--Animal-->dog

### real time example
* One company (parent) has multiple departments — Sales, Engineering, HR (children) —
* each inheriting general company policies but operating with their own distinct rules and tools.

### example


```python
class Animal:
    def eat(self):
        print("Animal eats")

class Dog(Animal):   # Create a Dog class that inherits from Animal
    def bark(self):   # defines the bark method
        print("Dog barks")

class Cat(Animal):      # Create a cat class that inherits from Animal
    def meow(self):     # defines the meow method
        print("Cat meows")
dog = Dog()
cat = Cat()

dog.eat()
dog.bark()

cat.eat()
cat.meow()
```

# 16.Polymorphism
* Polymorphism means the same operation can behave differently depending on the object.

### real time example
* Pressing "play" behaves differently depending on the device  a TV shows video, a speaker plays audio, a game console launches a game.
*  Same action name, different underlying behavior per object.

### example


```python
class Dog:
    def sound(self):
        print("Dog barks")
        
class Cat:
    def sound(self):
        print("Cat meows")
animals = [Dog(), Cat()]  # Create objects of Dog and Cat classes

for animal in animals:    # Loop through each animal object
    animal.sound()
```

# 17.Method overriding


* Method overriding happens when a child class redefines a method that already exists in its parent, replacing the parent's version specifically for that child's objects.
*  It's the mechanism that makes runtime polymorphism work

### real time example

* A general company policy says work hours: 9-5
*  but the Sales department overrides it with "work hours: 10-6" for their specific team,
*   replacing the inherited rule with their own version.

### example


```python
class Animal:
    def sound(self):         # Define the sound method
        print("Animal makes sound")

class Dog(Animal):
    def sound(self):     # Override the sound method
        print("Dog barks")
dog = Dog()
dog.sound()     # Call the overridden sound method
```

# 18.Method Overloading

* Method overloading means using the same method name with different numbers or types of parameters.
* Python does not support traditional method overloading in the same way we can use default arguments.

### real time example

* A single vending machine slot accepts coins, cards, or phone taps
*  one interface point handling multiple "kinds" of input, rather than three separate physical slots

### example

```python
class Calculator:
    def add(self, *numbers):   # Define add method to accept multiple numbers
        total = 0                # Initialize the total value
        for number in numbers:
            total += number
        return total
calculator = Calculator()   # Create an object of Calculator class
print(calculator.add(10, 20))  # Add two numbers
print(calculator.add(10, 20, 30))
print(calculator.add(10, 20, 30, 40))
```

# 19.Composition
* Composition means one object contains another object as a part of itself.
* It represents a strong has-a relationship.

### real time example
* A car "has-a" engine.
* The engine is built specifically for that car and doesn't meaningfully exist independently of it
*  if you scrap the car, that particular engine instance goes with it.

### example

```python
class Engine:
    def start(self):
        print("Engine starts")
        
class Car:
    def __init__(self):
        self.engine = Engine()
    def start_car(self):
        self.engine.start()
        print("Car starts")
car = Car()
car.start_car()
```

# 20.Aggregation
* Aggregation is a weaker "has-a" relationship than composition
* the contained object can exist independently and can be shared across multiple owners.
*  The part is passed in from outside rather than created internally.

### real time example

* A university "has" students, but students exist independently of any one university
*  they can transfer, and the same student concept isn't destroyed if a university closes.
*  The relationship is looser than a car and its engine.

### example

```python
class Teacher:
    def __init__(self, name):
        self.name = name

class Department:
    def __init__(self, teacher):
        self.teacher = teacher

teacher = Teacher("Ravi")
department = Department(teacher)
print(department.teacher.name)
```

# 21. Association
* Association means two independent objects are connected or interact with each other.
* It does not necessarily mean one object owns the other.

### real time example
* A student and teacher interact.
* The student can exist without the teacher, and the teacher can exist without the student.

### example

```python
class Student:
    def study(self):
        print("Student is studying")
class Teacher:
    def teach(self, student):
        print("Teacher is teaching")
        student.study()

student = Student()
teacher = Teacher()
teacher.teach(student)
```

# 23.super()
* super() is used to access functionality from a parent class.
* It is especially useful when a child class wants to extend the parent's constructor or method.

# real time usecase
* A new manager still follows the company's general onboarding process (parent) before adding department specific training (child)
* super() is like saying "first, do everything the standard onboarding does, then I'll add my own extra steps."

### example

```python
class Student:  
    def __init__(self, name):  # Constructor to initialize the name
        self.name = name  

class CollegeStudent(Student):  # Creates a child class  inherited  from Student
    def __init__(self, name, course):  # Constructor to initialize name and course
        super().__init__(name)  # Call the parent class constructor
        self.course = course  
student = CollegeStudent("Ravi", "B.Tech")  # Create an object of CollegeStudent

print(student.name)  
print(student.course) 
```

# 24.\_\_str\_\_\()

* defines a user-friendly string representation of an object

### real time example
* Instead of displaying an object's default memory-related representation, we can display meaningful information.

### example

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):  # Define the __str__ method to display object details
        return f"Student: {self.name}, Marks: {self.marks}"

student = Student("Ravi", 85)

print(student)
```

# 25.\_\_repr_\_\()

* provides a more detailed representation of an object, mainly useful for developers and debugging.
* A common goal is to make the representation unambiguous

### real time example

* We can represent a Student object in a form that clearly shows its data.

### example

```python
class Student:

    def __init__(self, name):
        self.name = name

    def __repr__(self): # defining the __repr__ function
        return f"Student('{self.name}')"
student = Student("sridhar")
print(repr(student))
```

# 26.\_\_len_\_\()

* defines what should happen when Python's len() function is used with an object.

### real time example

* Asking "how many items are there in this shopping cart?"
* should work the same way regardless of whether it's a physical cart or a custom

### example

```python
class Student:

    def __init__(self, subjects):
        self.subjects = subjects
    def __len__(self):
        return len(self.subjects)
student = Student(["Python", "SQL", "Java"])

print(len(student)) # prints no.of subjects is there in the list
```

# 27.Dataclasses

* A dataclass is a convenient way to create classes that mainly store data.

* Python can automatically generate methods such as __init__() and __repr__().

### real time example

* Filling out a pre-printed form (name, age, email fields already laid out) is much faster than designing a blank form from scratch every time

#### example

```python
from dataclasses import dataclass

@dataclass      # Automatically provides useful methods for the class
class Student:
    name: str
    age: int
    marks: float
student = Student("Ravi", 20, 85.5)  # Create a Student object with values

print(student)
```

# 28.Singleton 

* The Singleton Pattern is used when we want a class to have only one instance during the lifetime of the program.

### real time example

* Imagine an application configuration manager.
* Instead of creating multiple configuration objects, we want one shared configuration object.

### example

```python
class Singleton:

    _instance = None 
    def __new__(cls):
        if cls._instance is None:   # Check if an object already exists
            cls._instance = super().__new__(cls) # Create the object only once
        return cls._instance

object1 = Singleton()  # Create the first object
object2 = Singleton()  # Try to create another object

print(object1 is object2)
```
