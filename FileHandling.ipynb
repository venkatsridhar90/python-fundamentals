# File Handling

## Reading text files
* Reading a text file means opening an existing file and getting the information stored inside it.
* Python provides the open() function to work with files. We can use "r" mode to read a file.

### real time example
* imagine a teacher has taken a attendence in a file, if we open the file the file we can just read the attendence information

### example

```python
file = open("students.txt", "r")        # Open the file in read mode
content = file.read()                   # Read the complete content
print(content)                         # Display the content
file.close()                           # Close the file
```

## Writing text files
* Writing text file means storing information inside a file using Python.
* We can use "w" mode to write data. If the file does not exist, Python creates it.
### real time example
* Imagine a teacher wants to save student names in a file.
* Instead of writing the names manually, Python can create the file and store them.
### example

```python
file = open("students.txt", "w")  # Open the file in write mode
file.write("Welcome to python programming\n")
file.write("This file contains python Programs\n")
file.write("welcome to hello world\n")

file.close()

print("Student names saved successfully")
```

## CSV files
* CSV stands for Comma Separated Values
* It stores the information in the form of rows and columns
* It is like excel table
### real time example
* The college has a student database which contains of
* Student Name
* StudentId
* StudentAge
### example

```python
import csv      # imports the csv module
with open("students.csv", "w", newline="") as file:   # Open CSV file in write mode
    writer = csv.writer(file)           # Create a CSV writer
    writer.writerow(["Name", "Age", "Marks"]) # column headings

    writer.writerow(["Ravi", 20, 85])      # student records
    writer.writerow(["Sai", 21, 90])
    writer.writerow(["Kiran", 20, 78])
print("CSV file created ")
```

## JSON Files
* JSON stands for java Script Object Notation
* It is commonly used for storing and exchanging structured data.
* Python dictionaries and JSON objects have a similar structure.
### real time example
* imagine a student id card contains information like
* name - Sridhar
* date of birth - 01/09/2005
* course - python
### example


```python
import json              # immports the json module
student = {
    "name": "sridhar",
    "age": 21,
    "course": "python"
}

with open("student.json", "w") as file:         # opens a json file with write mode
    json.dump(student, file, indent=4)
print("JSON file created")
```

## Pickle files
* Pickle is used to save Python objects into a file and later load them back.
*  Pickle can store Python objects such as:
* Lists
* Dictionaries
* Tuples
* Objects
* Machine learning models
### real time example
* saving a photo album if we go to a trip instead of keeping them separate
* we can create an album and we can see whenever we want
### example

```python
import pickle      # import the pickle module 

student = {
    "name": "Sridhar",
    "college": "SRM UNIVERSITY"
}

with open("student.pkl", "wb") as file:   # opens the file with write-binary mode
    pickle.dump(student, file)
print("Object saved")
```

## Context Manager(with)
* It makes sure that the resource is properly closed or cleaned up after the work is finished
* Simply we can say that it handles with opening and closing resource,so we don't have to manage it manually
### real time example
* imagine taking a book from library resouce is opened
* reading the book using the resource
* returning the book resource is completed
### example

```python
with open("message.txt", "w") as file: # Open file using context manager
    file.write("Welcome to Python")

print("File operation completed")
```

## pathlib
* provides an easy and modern way to work with file and folder paths.
* Instead of manually creating path strings, we can use Path.
### real time example
* we have a folder called students and we want to access it
* pathlib helps to create and manage the paths easily
### example

```python
from pathlib import Path    # path is imported from pathlib
file_path = Path("students") / "marks.txt"  # Create a path

print(file_path) # Display the path
print(file_path.exists()) # Check whether file exists
```

## OS
* It allows python to interact with the operating system
* with the help of os module we can check the file/folders, we can rename / remove the file
### real time example
*  imagine a house i need to create a new room and last new room was created

### example

```python
import os     # importing os module
current_folder = os.getcwd()       # Get current working directory

print("Current folder:", current_folder)   # Display current folder
     
print(os.listdir())     # Display files and folders
```

## File Validation
* checking whether a file exists or not and checks it is suitable for this operation or not before using it
### real time example
* Imagine a website where users can update profile picture
* the application checks the files whether it has correct extension or not
### example

```python
import os
filename = input("Enter the file name: ")     # user needs to enter the filename

if os.path.isfile(filename):
    print("File exists")
else:
    print("File does not exist")
```

## Directory Operations
* It is used to organise the files
* It contains operations like creating,renaming,checking,removing folders using python
### real time example
* a company can create folders like employees,projects,reports and salary
### example


```python
from pathlib import Path

folder = Path("new_folder")
if folder.exists():
    folder.rmdir()   # rempoves the file if it is empty
print("removed Successfully")
```
