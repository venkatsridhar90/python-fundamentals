### Recursive functions

* A recursive function is a function that calls itself in order to solve a problem
* a condition where it stops calling itself and returns a direct answer

## Real time example

* A file system has folders inside folders inside folders
* Recursion naturally handles this by calling itself on each subfolder until there are no more subfolders left.

## Example


```python
def fact(n):
    if n <= 1:            # base case
        return 1
    return n * fact(n - 1)   # recursive case
print(fact(7))   # 7 * 6 * 5 * 4 * 3 * 2 * 1 = 120
```

## Generators

* A generator produces values one at a time instead of creating all the values at once.
* This can be useful when working with large amounts of data.

## Real time example

* Instead of baking 1000 cookies and for storing it takes more space 
* so the bakery bakes one cookie at a time only when someone asks for the next one it saves efoort and storage

```python
def count(n):
    i = 1
    while i <= n:
        yield i        # pause here, return i, remember state
        i += 1
counter = count(4)
print(next(counter))   
print(next(counter))   
print(next(counter)) 
print(next(counter))
```

## Yield

* this keyword is used in generator to to return one value at a time
* it resumes from where it stopped without starting from first

### Real time example

* for example if we watching a web series or movies in ott apps and stopped in middle
* if we want to see that tomorrow it resumes from where we stopped.
  

### Example

```python
def count():
    yield 1        # gives 1, then pauses here
    yield 2        # gives 2, then pauses here
    yield 3        # gives 3, then pauses here

for num in count():   # runs the function and gets values one by one
    print(num)                  # prints 1, then 2, then 3
```

## Decorators

* decorator is used to add extra functionality to an existing function without changing the original function's code.

## Real world example

* a restaurant bill automatically adding a service charge without changing the total bill


## Example

```python
def decorator(func):
    def wrapper(a, b):
        print("Adding numbers...")
        func(a, b)
    return wrapper
@decorator # without changing the above function we can add the extra function
def add(a, b):
    print(a + b)

add(10, 20)
```

# Closure

* A closure happens when an inner function remembers and uses a value from its outer function even after the outer function has finished.

## Real time example

* sealed envelope: someone writes a note inside an envelope and seals it before handing to you. 
* Even though they've left the room , the envelope still "remembers" what was written inside the envelope

## Example

```python
def create_greet(name):
    def greet():
        print("Hello", name)
    return greet
my_greet = create_greet("Sai") # Call create_greet() and store the returned function
my_greet() # Call the returned function
```

## Global

* modify a variable defined outside the function, at the top level of the program , instead of creating a new local variable with the same name.

## Real time example

* school wide announcement,
* if the school management made an announcement that tomorrow is holiday it is applicable for the whole students and classes

## Example

```python
total_sales = 0

def add_sale(amount):
    global total_sales
    total_sales += amount    # updates the shared value, not a local copy

add_sale(500)  
add_sale(300)
print(total_sales)  
```

## Non local

* is used inside a nested function to modify a variable that belongs to an outer  function
*  tells Python to find and update the variable in the nearest outer function without creating a new variable with the same name 

## Real time example

* for class announcement
* it means if there is a test for class X students means only test for X students not for other classes.


## Example

```python
def outer():
    x = 10
    def inner():
        nonlocal x     # refers to outer's 'x', not global, not a new local
        x += 5
    inner()
    return x
print(outer())   
```
